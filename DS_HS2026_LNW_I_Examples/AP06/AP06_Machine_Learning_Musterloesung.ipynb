### B1 (8 Punkte)

**Ziel:** Ein multiples lineares Regressionsmodell (OLS, mit Konstante) und ein Random Forest Regression Model (100 Trees, `max_depth=5`) für `price_chf` sollen **fair verglichen** werden: Beide werden mit den vorbereiteten Trainingsdaten `X_train`, `y_train` trainiert und auf den **Testdaten** mit **R²** und **RMSE** (in CHF) beurteilt.

```python
# Fit the multiple linear regression model (OLS)
olsres_B1 = sm.OLS(y_train, sm.add_constant(X_train)).fit()

# Create and train the random forest regressor
reg_rf_B1 = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
reg_rf_B1.fit(X_train, y_train)

# Predict the prices of the random forest for the test data
y_pred_rf_B1 = reg_rf_B1.predict(X_test)

# Compare the models
print('OLS           R-squared:', round(olsres_B1.rsquared, 3),
      '| RMSE (CHF):', round(np.mean(olsres_B1.resid ** 2), 2))
print('Random Forest R-squared:', round(r2_score(y_test, y_pred_rf_B1), 3),
      '| RMSE (CHF):', round(np.mean((y_test - y_pred_rf_B1) ** 2), 2))
```

Tatsächliches Ergebnis:

```
OLS           R-squared: 0.966 | RMSE (CHF): 30840.79
Random Forest R-squared: 0.957 | RMSE (CHF): 39594.98
```

Der Analyst meldet: *«OLS ist besser. Aber mit einem Fehler von über CHF 30'000 pro Velo sind beide Modelle unbrauchbar.»*

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben). Begründen Sie jede Änderung gegenüber dem ursprünglichen Code und beurteilen Sie die Aussage des Analysten mit den korrigierten Kennzahlen.

**Begründung der Änderungen:**

1. **OLS auf den Testdaten auswerten:** `olsres_B1.rsquared` und `olsres_B1.resid` beziehen sich auf die **Trainingsdaten**, der Random Forest wird dagegen auf den **Testdaten** beurteilt — der Vergleich ist nicht fair (Trainingsfehler sind tendenziell zu optimistisch). Für OLS müssen die Preise für `X_test` vorhergesagt werden: `olsres_B1.predict(sm.add_constant(X_test))` — die Konstante muss auch bei den Testdaten ergänzt werden, sonst passen Features und Koeffizienten nicht zusammen. Anschliessend `r2_score(y_test, ...)`.
2. **Wurzel ziehen:** `np.mean((y - ŷ) ** 2)` ist der **MSE** (mittlerer quadrierter Fehler, Einheit CHF²), nicht der RMSE. Erst `np.sqrt(...)` liefert den RMSE in CHF.

Ergebnis auf den Testdaten: **OLS R² = 0.965, RMSE = CHF 181.11**; **Random Forest R² = 0.957, RMSE = CHF 198.98**.

Beurteilung: OLS ist tatsächlich etwas besser — aber erst der korrigierte Vergleich zeigt das zuverlässig. Plausibel, weil der Preis in den Daten weitgehend linear von den Features abhängt. Mit einem typischen Fehler von rund **CHF 180–200** sind beide Modelle keineswegs unbrauchbar.

**Korrigierter Code:**

````python
# Fit the multiple linear regression model (OLS)
olsres_B1 = sm.OLS(y_train, sm.add_constant(X_train)).fit()

# Predict the prices of the OLS model for the test data (note the 'add_constant' function)
y_pred_ols_B1 = olsres_B1.predict(sm.add_constant(X_test))

# Create and train the random forest regressor
reg_rf_B1 = RandomForestRegressor(n_estimators=100, max_depth=5, random_state=42)
reg_rf_B1.fit(X_train, y_train)

# Predict the prices of the random forest for the test data
y_pred_rf_B1 = reg_rf_B1.predict(X_test)

# Compare the models on the test data (RMSE = square root of the MSE)
print('OLS           R-squared:', round(r2_score(y_test, y_pred_ols_B1), 3),
      '| RMSE (CHF):', round(np.sqrt(np.mean((y_test - y_pred_ols_B1) ** 2)), 2))
print('Random Forest R-squared:', round(r2_score(y_test, y_pred_rf_B1), 3),
      '| RMSE (CHF):', round(np.sqrt(np.mean((y_test - y_pred_rf_B1) ** 2)), 2))
````

```
Accuracy: 0.733

```

Der Analyst folgert: *«Mit Brems- und Kettenverschleiss lässt sich der Servicebedarf kaum vorhersagen — das Modell ist kaum besser als ein Modell, das immer ‹kein Service› sagt (Accuracy 0.754).»*

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben). Begründen Sie jede Änderung gegenüber dem ursprünglichen Code.

**Begründung der Änderungen:**

1. **Richtige Features verwenden:** `X_train` und `X_test` enthalten die **Preis-Features** (`age_years`, `mileage_km`, `equipment_level`, `is_ebike`) aus dem Regressions-Split. Die Formel `cm[1, 1] / (cm[1, 0] + cm[1, 1])` berechnet dann TP / (alle vorhergesagt positiven) = **Precision** statt Recall. Richtig: `confusion_matrix(y_test_class, y_pred_B2)`.
