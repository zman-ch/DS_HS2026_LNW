````python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import r2_score, accuracy_score, confusion_matrix, RocCurveDisplay

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
````

### Laptopdaten laden → `df`

````python
# Import data from csv to pandas dataframe named 'df'
df = pd.read_csv('laptop_data.csv')

# Show dimensions and first rows
print('Number of rows and columns:', df.shape)
df.head()
````

### Trainings- und Testdaten erstellen

Der folgende Code teilt die Daten **zweimal** in 80 % Trainingsdaten und 20 % Testdaten auf:

- **Preis (Regression):** `X_train`, `X_test`, `y_train`, `y_test` mit den Features in `price_features` und der Zielvariable `price_chf`.
- **Rückgabe (Klassifikation):** `X_train_class`, `X_test_class`, `y_train_class`, `y_test_class` mit den Features in `return_features` und der Zielvariable `returned`. `stratify` sorgt dafür, dass der Anteil der zurückgegebenen Laptops in Trainings- und Testdaten gleich gross ist.

````python
# Train and test data (regression)
price_features = ['age_years', 'ram_gb', 'storage_gb', 'cpu_level', 'is_business']
X_train, X_test, y_train, y_test = train_test_split(df[price_features],
                                                    df['price_chf'],
                                                    test_size=0.20,
                                                    random_state=42)

# Train and test data (classification)
return_features = ['battery_health_pct', 'cosmetic_grade', 'charge_cycles']
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(df[return_features],
                                                                            df['returned'],
                                                                            test_size=0.20,
                                                                            random_state=42,
                                                                            stratify=df['returned'])

# Show the size of the samples
print('Regression:     ', X_train.shape[0], 'Trainingsdaten /', X_test.shape[0], 'Testdaten')
print('Klassifikation: ', X_train_class.shape[0], 'Trainingsdaten /', X_test_class.shape[0], 'Testdaten')
````

### B1 (8 Punkte)

**Ziel:** Ein Regression Tree (`max_depth=6`, `random_state=42`) für `price_chf` mit **allen fünf Preis-Features** gemäss Hinweisen zu den Daten — trainiert auf 80 % der Daten und beurteilt mit dem R² auf den 20 % Testdaten.

```python
# Create training and test data
X_train_B1, X_test_B1 = train_test_split(df[['age_years', 'ram_gb', 'storage_gb', 'is_business']],
                                         test_size=0.20,
                                         random_state=42)
y_train_B1, y_test_B1 = train_test_split(df['price_chf'],
                                         test_size=0.20,
                                         random_state=0)

# Create and train the regression tree
reg_tree_B1 = DecisionTreeRegressor(max_depth=6, random_state=42)
reg_tree_B1.fit(X_train_B1, y_train_B1)

# Predict the prices for the test data
y_pred_B1 = reg_tree_B1.predict(X_test_B1)

# Evaluate the model on the test data
print('R-squared (test data):', round(r2_score(y_test_B1, y_pred_B1), 3))
```

Tatsächliches Ergebnis:

```
R-squared (test data): -0.293
```

Der Analyst meldet: *«Das Modell ist schlechter als einfach immer den Durchschnittspreis zu schätzen — Trees funktionieren für Laptops nicht.»*

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben). Begründen Sie jede Änderung gegenüber dem ursprünglichen Code und erklären Sie, was ein negatives R² bedeutet.

**Begründung der Änderungen:**

1. **Features und Zielvariable gemeinsam aufteilen:** Features und Preise werden in zwei **getrennten** Aufrufen von `train_test_split()` mit verschiedenen `random_state` aufgeteilt. Dadurch landen in `X_train_B1` und `y_train_B1` **andere Laptops** — jedem Laptop wird der Preis eines zufälligen anderen Laptops zugeordnet. Das Modell lernt einen Zusammenhang, den es nicht gibt. Richtig: `train_test_split(X, y, ...)` in **einem** Aufruf, damit die Zeilen zusammenbleiben.
2. **Fehlendes Feature ergänzen:** `cpu_level` fehlt in der Feature-Liste; gemäss Ziel sollen alle fünf Preis-Features verwendet werden.

Ein **negatives R²** bedeutet, dass die Vorhersagen auf den Testdaten **schlechter** sind als die konstante Schätzung «Mittelwert von y» (R² = 0): Die Summe der quadrierten Fehler ist grösser als die Gesamtvariabilität.

Ergebnis: R² auf den Testdaten **0.878** — der Tree erklärt rund 88 % der Variabilität der Preise.

**Korrigierter Code:**

````python
# Create training and test data (features and target in ONE call, all five price features)
X_train_B1, X_test_B1, y_train_B1, y_test_B1 = train_test_split(df[['age_years', 'ram_gb', 'storage_gb', 'cpu_level', 'is_business']],
                                                                df['price_chf'],
                                                                test_size=0.20,
                                                                random_state=42)

# Create and train the regression tree
reg_tree_B1 = DecisionTreeRegressor(max_depth=6, random_state=42)
reg_tree_B1.fit(X_train_B1, y_train_B1)

# Predict the prices for the test data
y_pred_B1 = reg_tree_B1.predict(X_test_B1)

# Evaluate the model on the test data
print('R-squared (test data):', round(r2_score(y_test_B1, y_pred_B1), 3))
````

>

### B2 (8 Punkte)

**Ziel:** Ein **Random-Forest-Klassifikationsmodell** (100 Trees, `random_state=42`) sagt `returned` vorher. Beurteilt wird es auf den Testdaten mit Accuracy, Confusion Matrix und der ROC-Kurve inkl. AUC.

```python
# Create and train the random forest model
rf_B2 = RandomForestRegressor(n_estimators=100, random_state=42)
rf_B2.fit(X_train_class, y_train_class)

# Predict the target variable for the test data
y_pred_B2 = rf_B2.predict(X_test_class).round()

# Evaluate the model on the test data
print('Accuracy:', round(accuracy_score(y_test_class, y_pred_B2), 3))
print('Confusion matrix:')
print(confusion_matrix(y_test_class, y_pred_B2))

roc_B2 = RocCurveDisplay.from_predictions(y_test_class, y_pred_B2)

```

Tatsächliches Ergebnis:

```
Accuracy: 0.835
Confusion matrix:
[[124  17]
 [ 16  43]]
AUC: 0.8
```

Die ROC-Kurve besteht nur aus zwei geraden Linienstücken.

**Aufgabe:** Schreiben Sie den korrigierten Code in die Zelle «Korrigierter Code» (die Kommentare des ursprünglichen Codes sind als Gerüst vorgegeben). Begründen Sie jede Änderung gegenüber dem ursprünglichen Code und erklären Sie, warum die ROC-Kurve nur aus zwei Linienstücken besteht.

**Begründung der Änderungen:**

1. **Klassifikations- statt Regressionsmodell:** `returned` ist eine Klasse (0/1) — es braucht einen `RandomForestClassifier`. Der `RandomForestRegressor` minimiert den quadrierten Fehler und liefert Durchschnittswerte, die erst durch `.round()` zu 0/1 «gemacht» werden. Der Classifier wählt die Splits mit dem Gini-Kriterium und liefert über `predict()` direkt die Klasse bzw. über `predict_proba()` die Wahrscheinlichkeit.
2. **ROC-Kurve aus Wahrscheinlichkeiten:** Die ROC-Kurve entsteht, indem für **jeden** Schwellenwert τ die Falsch- und Richtig-Positiv-Rate berechnet wird. Dafür braucht es kontinuierliche Scores (Wahrscheinlichkeiten). Mit gerundeten 0/1-Vorhersagen gibt es nur **einen** sinnvollen Schwellenwert — die Kurve besteht deshalb nur aus einem Punkt, verbunden mit (0, 0) und (1, 1), und die AUC wird unterschätzt. Richtig: `RocCurveDisplay.from_estimator(rfc_B2, X_test_class, y_test_class)` (verwendet intern `predict_proba`).

Ergebnis: Accuracy **0.865**, Confusion Matrix `[[128 13] [14 45]]`, **AUC 0.89** — das Modell unterscheidet zurückgegebene und behaltene Laptops gut.

**Korrigierter Code:**

````python
# Create and train the random forest model (classifier instead of regressor)
rfc_B2 = RandomForestClassifier(n_estimators=100, random_state=42)
rfc_B2.fit(X_train_class, y_train_class)

# Predict the target variable for the test data
y_pred_B2 = rfc_B2.predict(X_test_class)

# Evaluate the model on the test data
print('Accuracy:', round(accuracy_score(y_test_class, y_pred_B2), 3))
print('Confusion matrix:')
print(confusion_matrix(y_test_class, y_pred_B2))

# ROC curve and AUC (based on the predicted probabilities)
roc_B2 = RocCurveDisplay.from_estimator(rfc_B2, X_test_class, y_test_class)
plt.show()
print('AUC:', round(roc_B2.roc_auc, 2))
````
